```python
from pymongo import MongoClient

# Connect to local MongoDB
client = MongoClient('mongodb://localhost:27017/')

# Select the database
db = client['info_demo']

# Select the collection
collection = db['info']

# Document to insert
document = {"name": "sneha", "city": "Mumbai"}

# Insert the document
inserted_doc = collection.insert_one(document)

# Print the inserted ID
print("Inserted document ID:", inserted_doc.inserted_id)
```
